```python
# Create and Print a Dictionary
student = {
    "name": "maheswari",
    "age": 20,
    "course": "Python-fullstack"
}

print(student)
```

```python
# Access a Value Using Key
student = {
    "name": "Maheswari",
    "age": 20,
    "course": "Python-fullstack"
}

print(student["name"])
```

```python
# Add a New Key-Value Pair
student = {
    "name": "Maheswari",
    "age": 20,
}

student["course"] = "Python-fullstack"

print(student)
```

```python
# Update an Existing Value
student = {
    "name": "Maheswari",
    "age": 20,
}

student["age"] = 21

print(student)
```

```python
# Delete a Key-Value Pair
student = {
    "name": "Maheswari",
    "age": 20,
    "course": "Python-fullstack"
}

del student["course"]

print(student)
```

```python
# Check if a Key Exists
student = {
    "name": "Maheswari",
    "age": 20,
    "course": "Python-fullstack"
}

key = input("Enter key: ")

if key in student:
    print("Key exists")
else:
    print("Key does not exist")
```

```python
# Print all keys
student = {
    "name": "Maheswari",
    "age": 20,
    "course": "Python-fullstack"
}

print(student.keys())
```

```python
# Print all values
student = {
    "name": "Maheswari",
    "age": 20,
    "course": "Python-fullstack"
}

print(student.values())
```

```python
# Print all key-value pairs
student = {
    "name": "Maheswari",
    "age": 20,
    "course": "Python-fullstack"
}

for key, value in student.items():
    print(key, ":", value)
```
