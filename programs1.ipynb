```python
# Check Palindrome Words in a List

def check_palindrome(lst):
    for word in lst:
        if word == word[::-1]:
            print(word, "Palindrome")
        else:
            print(word, "Not Palindrome")

lst = input("Enter words: ").split(",")
check_palindrome(lst)
```

```python
# Check Leap Years in a List

def check_leap(years):
    for year in years:
        if year % 4 == 0 and (year % 100 != 0 or year % 400 == 0):
            print(year, "Leap Year")
        else:
            print(year, "Not a Leap Year")

years = list(map(int, input("Enter years: ").split(",")))
check_leap(years)
```

```python
def prime(lst):
    for n in lst:
        if n < 2:
            print(n, "Not Prime")
        else:
            for i in range(2, n):
                if n % i == 0:
                    print(n, "Not Prime")
                    break
            else:
                print(n, "Prime")

lst = list(map(int, input("Enter numbers: ").split(",")))
prime(lst)
```

```python
def fib(num):
    if num <= 1:
        return num
    return fib(num - 1) + fib(num - 2)

n = int(input("Enter number of terms: "))

for i in range(n):
    print(fib(i))
```

```python
# Factorial of a number using Recursion

def factorial(num):
    if num <= 1:
        return 1
    return num * factorial(num - 1)

num = int(input("Enter a number: "))
result = factorial(num)

print("Factorial =", result)
```

```python
#patterns
def pattern(n):
    for i in range(1,5):
        for j in range(i):
            print(n,end=" ")
            n+=1
        print()
pattern(n=1)
```

```python
#amstrong
n = int(input("Enter a number: "))

s = 0
t = n

while t > 0:
    d = t % 10
    s = s + d**3
    t = t // 10

if s == n:
    print("Armstrong")
else:
    print("Not Armstrong")
```
